El problema a escala mundial. La FAO estima que cerca del 14 % de los alimentos del mundo se pierde entre la cosecha y la venta minorista, y que las pérdidas y el desperdicio son en general mayores en frutas y verduras que en cereales y legumbres. El PNUMA calculó además que en 2022 se generaron 1.050 millones de toneladas de alimentos desperdiciados, casi una quinta parte de lo disponible para el consumidor. Parte de ese desperdicio se explica porque los consumidores exigen una calidad estética, y porque los productos perecederos tienen poca vida útil. Por eso la clasificación temprana y objetiva de la calidad es relevante. 


El caso de Colombia. Según el estudio de la Dirección de Seguimiento y Evaluación de Políticas Públicas del DNP (2016), el país pierde o desperdicia 9,76 millones de toneladas de alimentos al año, y el 62 % corresponde a frutas y verduras. Del total, el 63,8 % de las pérdidas ocurre antes de la distribución. Esto sirve de argumento: la mayor parte de la pérdida ocurre en las etapas donde se clasifica y se maneja el producto (poscosecha, acopio, plazas de mercado y centrales de abasto), no en el hogar. Ahí es donde un sistema de clasificación automático podría aportar.

Por qué la clasificación manual es insuficiente. Bhargava y Bansal señalan que la clasificación humana es inconsistente, lenta, variable, subjetiva y fácilmente influenciada por el entorno. Una revisión de 2026 agrega que depende del juicio visual y táctil y se ve afectada por la habilidad, la fatiga y los criterios subjetivos del operario, lo que dificulta tomar decisiones iguales a ritmo rápido. Ismail y Malik describen la inspección visual manual como laboriosa y con inconsistencias en la inspección y la clasificación. 


Qué significa "calidad" y "tamaño" (útil para definir sus etiquetas). Los estándares de comercialización del tomate, por ejemplo, definen tres clases: Extra, I y II. El estándar Codex para tomates define una tabla de calibres por diámetro en milímetros. Esto les da una base normativa para justificar sus 3 o más categorías de calidad y sus rangos de tamaño, en vez de inventarlos. Puede adaptarse a otras frutas. 


Antecedentes técnicos. Ismail y Malik compararon ResNet, DenseNet, MobileNetV2, NASNet y EfficientNet para la clasificación de frutas. Nithya et al. aplicaron una CNN a la detección de defectos en mango. Amna et al. construyeron un prototipo con aprendizaje profundo y control mecánico, con exactitudes de validación cercanas al 94 % en mango y tomate. Esto permite posicionar su proyecto frente a la literatura y fijar una referencia realista de desempeño.
